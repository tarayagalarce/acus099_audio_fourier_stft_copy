## Objetivo
#### Analizar las señales de audio de biofonía, geofonía y antropofonía mediante la FFT Y STFT


## Qué hacen los códigos

Librosa y Soundfile: abren y preparan los archivos de audio 
Numpy: Almacena las muestras, representa las señales como arreglos multidimensionales
Scipy: Permite aislar frecuencias especificas, como ruido
Matplotlib: Dibuja graficas y espectrogramas
Ipython: Genera el reproductor para poder escuchar los audios
