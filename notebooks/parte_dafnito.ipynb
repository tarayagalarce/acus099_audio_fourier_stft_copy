from pathlib import Path




## Audios
audio_dir = Path("../data/audio_tracks")

track_files = {
    "Antro1": audio_dir / "Antropofonia_1_(MOTOSIERRA).wav",
    "Antro2": audio_dir / "Antropofonia_2_(PIEDRAS).wav",
    "Bio1": audio_dir / "Biofonia_1_(HYRAX).wav",
    "Bio2": audio_dir / "Biofonia_2_(GRILLOS).wav",
    "Geo1": audio_dir / "Geofonia_1_(lluvia).wav",
    "Geo2": audio_dir / "Geofonia_2_(viento).wav",
} #diccionario con los nombres de las pistas y sus rutas correspondientes

track_files 

```python
import numpy as np #necesitamos numpy para manejar los arrays de audio
import matplotlib.pyplot as plt # necesitamos matplotlib para graficar las señales de audio
import librosa #librosa es una biblioteca de Python para análisis de audio, la usaremos para cargar las pistas de audio
import soundfile as sf #soundfile es una biblioteca para leer y escribir archivos de audio, la usaremos para guardar los resultados
from scipy.signal import welch #welch es una función de scipy para estimar la densidad espectral de potencia, la usaremos para analizar el contenido frecuencial de las pistas
from IPython.display import Audio, display #estas funciones nos permiten reproducir audio directamente en el notebook
```

```python
#FRECUENCIA DE MUESTREO

target_sr = 44100 #definimos la frecuencia de muestreo a 44.1 kHz

tracks = {} #diccionario para almacenar las pistas de audio cargadas

for name, path in track_files.items():
    signal, sr = librosa.load(path, sr=target_sr, mono=True)
    tracks[name] = signal #almacenamos la señal de audio en el diccionario con su nombre como clave

print("Pistas cargadas correctamente.\n")  # mensaje de confirmación

for name, signal in tracks.items():
    duration = len(signal) / target_sr #calculamos la duración de la pista dividiendo el número de muestras por la frecuencia de muestreo
    print(f"{name:8s} | samples = {len(signal):9d} | duration = {duration:8.2f} s") #imprimimos el nombre de la pista, el número de muestras y la duración en segundos
```
